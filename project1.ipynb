## Target Variable Definition

## Categorical Encoding
Categorical variables were converted into numerical format using one-hot encoding.

## Train-Test Split
The dataset was split into training and testing sets to evaluate model performance and avoid overfitting.


```python
from sklearn.model_selection import train_test_split
x_train , x_test , y_train , y_test = train_test_split(
    x , y , test_size=0.2 , random_state=42 ,
    stratify=y
)
```

test_size=0.2 لما قلت

اقصد فيه 20% اختبار

و80% تدريب

random_state=42  يخلي النتائج ثابته كل مره

ومهم للتكرار والتقييم

stratify=y يحافظ على نسبه المتعثرين والمسددين

مهم جدا لان البيانات غير متوازيه

```python
print(y_train.value_counts(normalize=True))
print(y_test.value_counts(normalize=True))
```

نتحقق ومفروض تكون النسب متقاربه

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()

x_train_scaled = scaler.fit_transform(x_train)
x_test_scaled = scaler.transform(x_test)
#
```

#StandardScaler

هنا استخدم توحيد المقاييس لان بعض المتغيرات قيمها كبيره وبعضها صغيره

## Model 1: Logistic Regression
Logistic Regression was chosen for its simplicity and interpretability in credit risk prediction.


```python
from sklearn.linear_model import LogisticRegression

model = LogisticRegression(
    max_iter=1000,
    class_weight='balanced'

)
model.fit(x_train_scaled,y_train)
```

#LogisticRegression اول مودل



*   بسيط و قابل للتفسير
*   مناسب للbinary classification



```python
y_pred = model.predict(x_test_scaled)

y_proba = model.predict_proba(x_test_scaled)[:,1]
```

التوقع النهائي 1 او 0 y_pred

احتمال التعثر y_proba

```python
from sklearn.metrics import roc_auc_score , confusion_matrix , classification_report
print(confusion_matrix(y_test,y_pred))
print(classification_report(y_test,y_pred))

roc_auc = roc_auc_score(y_test,y_proba)
print("ROC AUC  : ",roc_auc)
```

## Model 2: Decision Tree Classifier
A Decision Tree model was trained to capture non-linear relationships in the data.

```python
from sklearn.tree import DecisionTreeClassifier
from sklearn.metrics import accuracy_score, classification_report

# تدريب النموذج
dt_model = DecisionTreeClassifier(
    max_depth=5,
    random_state=42
)

dt_model.fit(x_train, y_train)

# التنبؤ
y_pred_dt = dt_model.predict(x_test)

# التقييم
print("Decision Tree Accuracy:", accuracy_score(y_test, y_pred_dt))
print(classification_report(y_test, y_pred_dt))
```
